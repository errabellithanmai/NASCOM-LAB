```python
print("\n--- Task 1.1: Data Generation and Basic Structures ---")
import random
from datetime import datetime, timedelta

# 1. Define configuration constants
DEVICES = [f"DEV_{i:03d}" for i in range(1, 11)]
TYPES = ["Temperature", "Humidity", "Pressure", "Vibration"]
LOCATIONS = ["Lab-Main", "Warehouse-A", "AssemblyLine-3", "ServerRoom-1"]

# Base time for the simulation
start_time = datetime(2026, 7, 20, 12, 0, 0)
sensor_readings = []

# 2. Generate 100 entries using a for loop
for i in range(100):
    # Increment time randomly by 5 to 30 seconds per loop iteration
    start_time += timedelta(seconds=random.randint(5, 30))
    timestamp_str = start_time.strftime("%Y-%m-%d %H:%M:%S")

    # Select random attributes
    device_id = random.choice(DEVICES)
    device_type = random.choice(TYPES)
    location = random.choice(LOCATIONS)

    # Generate values based on the sensor type
    if device_type == "Temperature":
        value = round(random.uniform(18.0, 35.0), 2)  # in Celsius
    elif device_type == "Humidity":
        value = round(random.uniform(35.0, 65.0), 2)  # in %
    elif device_type == "Pressure":
        value = round(random.uniform(980.0, 1020.0), 2)  # in hPa
    else:  # Vibration
        value = round(random.uniform(0.1, 5.5), 2)  # in mm/s

    # Append the structured tuple to our list
    sensor_readings.append(
        (timestamp_str, device_id, device_type, location, value)
    )

# 3. Print the first 5 results to verify
for reading in sensor_readings[:5]:
    print(reading)
```

```python
print("\n--- Task 1.2 ---")
raw_sensor_data = [
    {"timestamp": "2026-07-20T14:00:00", "temperature": 32.5, "humidity": 65},
    {"timestamp": "2026-07-20T14:05:00", "temperature": 33.1, "humidity": 64},
    {"timestamp": "2026-07-20T14:10:00", "temperature": 32.8, "humidity": 66}
]

# Extracting only the readings
temperature_readings = [reading["temperature"] for reading in raw_sensor_data]

print(temperature_readings)
# Output: [32.5, 33.1, 32.8]
```

```python
print("\n--- Task 1.4: Device Type Statistics ---")

# Initialize the dictionary with sensible large/small values
# Using float('inf') for min ensures any real reading will be smaller
# Using float('-inf') for max ensures any real reading will be larger
device_stats = {
    "Temperature": (float('inf'), float('-inf')),
    "Humidity": (float('inf'), float('-inf')),
    "Pressure": (float('inf'), float('-inf')),
    "Vibration": (float('inf'), float('-inf'))
}

# Iterate through all generated sensor readings to calculate min and max
for reading in sensor_readings:
    device_type = reading[2]
    value = reading[4]

    # Get the current min and max values for this device type
    current_min, current_max = device_stats[device_type]

    # Update values if a new extreme is found
    new_min = min(current_min, value)
    new_max = max(current_max, value)

    # Save the updated tuple back to the dictionary
    device_stats[device_type] = (new_min, new_max)

# Print the final statistics dictionary
print(device_stats)
```

```python
print("\n--- Task 2: Function Definitions and Usage ---")

# Function to count readings of a specific device type
def count_device_type(sensor_readings, device_type):
    count = 0
    for reading in sensor_readings:
        if reading[2] == device_type:
            count += 1
    return count

# Example usage
temperature_count = count_device_type(sensor_readings, "Temperature")
print("Temperature readings:", temperature_count)
```
