```python
battery_data = [  (0, 32, 400, 20, 95),
                  (1, 35, 395, 22, 88),
                  (2, 38, 390, 25, 80),
                  (3, 40, 385, 28, 70),
                  (4, 42, 380, 30, 60),
                  (5, 45, 375, 32, 48) ]
print("Time Temperature Voltage Current Charge")
for reading in battery_data:
   print(reading)
```

```python
for reading in battery_data:
   temperature = reading[1]
   if temperature > 40:
    print(f"Time {reading[0]}: High Temperature Alert")
   else:
     print(f"Time {reading[0]}: Temperature Normal")
```

```python
for reading in battery_data:
   charge = reading[4]
   if charge < 50:
    print(f"Time {reading[0]}: Low Battery Alert")
   else:
    print(f"Time {reading[0]}: Battery Charge Normal")
```

```python
def battery_status(charge):
  if charge > 70:
     return "High"
  elif 50 <= charge <= 70:
    return "Medium"
  else:
    return "Low"
for reading in battery_data:
      charge = reading[4]
      print(f"Charge: {charge}% -> {battery_status(charge)}")
```

```python
import numpy as np
D= np.array([  [32, 400, 20, 95],
                 [35, 395, 22, 88],
                 [38, 390, 25, 80],
                 [40, 385, 28, 70],
                 [42, 380, 30, 60],
                 [45, 375, 32, 48] ])
print(D)
```

```python
averages = np.mean(D, axis=0)

print("Average Temperature:", averages[0])
print("Average Voltage:", averages[1])
print("Average Current:", averages[2])
print("Average Charge:", averages[3])
```

```python
import numpy as np
A = np.array([ [2, 1], [1, 3] ])
X = np.array([ [40], [25] ])
Y = A @ X
print(Y)
```

```python
A = np.array([ [2, 1], [1, 3] ])
B = np.array([ 100, 120 ])
solution = np.linalg.solve(A, B)
print(solution)
```
