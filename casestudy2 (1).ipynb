```python
import pandas as pd
data={
    'student':['s1','s2','s3','s4','s5','s6','s7','s8','s9','s10'],
    'study_hours':[2,3,4,5,6,7,8,4,6,9],
    'attendance':[65,70,75,80,85,90,92,78,88,95],
    'assignment_score':[55,60,65,72,78,85,88,70,80,92],
    'previous_score':[58,62,68,70,76,82,86,72,79,90],
    'final_score':[60,64,70,75,82,88,91,74,84,94]

}
df=pd.DataFrame(data)
print(df)
print("no. of students and columns in the dataframe:")
print(df.shape)
print("datatypes:")
print(df.dtypes)
print("first five records:")
print(df.head())
print("last five records:")
print(df.tail())

```

```python
print("average study hours:",df['study_hours'].mean())
print("average attendance:",df['attendance'].mean())
print("average assignment score:",df['assignment_score'].mean())
print("average previous score:",df['previous_score'].mean())
print("average final score:",df['final_score'].mean())
max=df['final_score'].max()
min=df['final_score'].min()
print("student with the highest final score:",df[df['final_score']==max])
print("student with the lowest final score:",df[df['final_score']==min])
print("student whose attendance is greater then 80%:",df[df['attendance']>80])
print("students whose final score is greater than 80:",df[df['final_score']>80])
print("statistical summary:",df.describe())
```

```python
final_score_array = df['final_score'].to_numpy()
print("\nFinal Score as NumPy Array:")
print(final_score_array)
import numpy as np
print("mean of final score array:",np.mean(final_score_array))
print("median of final score array:",np.median(final_score_array))
print("maximum of final score array:",np.max(final_score_array))
print("minimum of final score array:",np.min(final_score_array))
print("standardb derviation of final score array:",np.std(final_score_array))
```

```python
def F(x):
    return 40 + 5 * x - 0.2 * x**2

def dF(x):
    return 5 - 0.4 * x

print("x\tF(x)\tdF(x)")
print("----------------------")
for x in range(1, 11):
    print(f"{x}\t{F(x):.2f}\t{dF(x):.2f}")
```

```python
import numpy as np

def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 1001)
area = np.trapezoid(L(x), x)

print(area)
```

```python
import numpy as np
X_features = df[["study_hours", "attendance", "assignment_score"]].values
Y = df["final_score"].values
X = np.column_stack((np.ones(len(X_features)), X_features))
beta = np.linalg.inv(X.T @ X) @ X.T @ Y
print("Coefficients:")
print(f"b0 (Intercept): {beta[0]:.4f}")
print(f"b1 (Study Hours): {beta[1]:.4f}")
print(f"b2 (Attendance): {beta[2]:.4f}")
print(f"b3 (Assignment Score): {beta[3]:.4f}")
```
